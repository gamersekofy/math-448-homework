1. Each p-value tests the null hypothesis that the corresponding predictor has no linear association with sales, holding the other predictors constant.

    - Intercept: The null hypothesis is that expected sales are zero when TV, radio, and newspaper advertising spending are all zero. Since (p < 0.0001), we reject this hypothesis.
    - TV: The null hypothesis is that TV advertising spending is not associated with sales after accounting for radio and newspaper spending. Since (p < 0.0001), there is strong evidence that TV advertising is associated with sales. We reject this too.
    - Radio: The null hypothesis is that radio advertising spending is not associated with sales after accounting for TV and newspaper spending. Since (p < 0.0001), there is strong evidence that radio advertising is associated with sales. Once again, we reject.
    - Newspaper: The null hypothesis is that newspaper advertising spending is not associated with sales after accounting for TV and radio spending. Since (p = 0.8599), we fail to reject the null hypothesis. There is no significant evidence that newspaper advertising is associated with sales in this model. We don't reject.

```python
# 10
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from ISLP import load_data

Carseats = load_data("Carseats")

# (a) Full model
full_model = smf.ols("Sales ~ Price + C(Urban) + C(US)", data=Carseats).fit()
print(full_model.summary())

# (d) Coefficient tests
print("\nP-values:")
print(full_model.pvalues)

# (e) Reduced model: retain Price and US
reduced_model = smf.ols("Sales ~ Price + C(US)", data=Carseats).fit()
print("\nReduced model:")
print(reduced_model.summary())

# (f) Compare model fit
print("\nModel comparison:")
print(pd.DataFrame({
    "Model": ["Full", "Reduced"],
    "R-squared": [full_model.rsquared, reduced_model.rsquared],
    "Adjusted R-squared": [
        full_model.rsquared_adj,
        reduced_model.rsquared_adj
    ],
    "Residual standard error": [
        np.sqrt(full_model.mse_resid),
        np.sqrt(reduced_model.mse_resid)
    ]
}))

# (g) 95% confidence intervals for the reduced model
print("\n95% confidence intervals:")
print(reduced_model.conf_int(alpha=0.05))

# (h) Outlier and leverage diagnostics
influence = reduced_model.get_influence()
summary_frame = influence.summary_frame()

n = len(Carseats)
p = int(reduced_model.df_model) + 1
leverage_cutoff = 2 * p / n

diagnostics = pd.DataFrame({
    "studentized_residual": summary_frame["student_resid"],
    "leverage": summary_frame["hat_diag"],
    "cooks_distance": summary_frame["cooks_d"]
})

print("\nPotential outliers:")
print(diagnostics[diagnostics["studentized_residual"].abs() > 2])

print("\nPotential high-leverage observations:")
print(diagnostics[diagnostics["leverage"] > leverage_cutoff])

# Diagnostic plots
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].scatter(reduced_model.fittedvalues, reduced_model.resid)
axes[0].axhline(0, color="red", linestyle="--")
axes[0].set_xlabel("Fitted values")
axes[0].set_ylabel("Residuals")
axes[0].set_title("Residuals vs. Fitted Values")

sm = influence.summary_frame()
axes[1].scatter(np.arange(n), sm["cooks_d"])
axes[1].set_xlabel("Observation")
axes[1].set_ylabel("Cook's distance")
axes[1].set_title("Cook's Distance")

plt.tight_layout()
plt.show()

# (h) The residual plot shows observations with studentized residuals
# greater than 2 in absolute value, which means there are possible outliers. However,
# the residuals are generally centered around zero and show no strong pattern.
# Many observations have relatively large Cook's distances compared with
# the 4/n guideline, but all are far below 1, so none appears highly
# influential.
```

```python
# 13
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Use the default random number generator with seed 1.
def generate_data(error_sd):
    rng = np.random.default_rng(1)
    x = rng.normal(loc=0, scale=1, size=100)
    eps = rng.normal(loc=0, scale=error_sd, size=100)
    y = -1 + 0.5 * x + eps
    return x, eps, y

# (a)-(c): original data, where Var(eps) = 0.25.
x, eps, y = generate_data(error_sd=0.5)
print(f"(c) Length of y: {len(y)}")
print("    beta_0 = -1 and beta_1 = 0.5")

# (d) Scatterplot.
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(x, y, alpha=0.75)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Original data: y versus x")
plt.show()

# (e) Fit the least-squares model.
X = sm.add_constant(x)
original_model = sm.OLS(y, X).fit()
print("\n(e) Original least-squares estimates:")
print(original_model.summary())
print(f"Estimated beta_0 = {original_model.params[0]:.4f}")
print(f"Estimated beta_1 = {original_model.params[1]:.4f}")

# (f) Least-squares line and population regression line.
x_grid = np.linspace(x.min(), x.max(), 200)
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(x, y, alpha=0.6, label="Observed data")
ax.plot(x_grid, original_model.params[0] + original_model.params[1] * x_grid,
        color="red", label="Least-squares line")
ax.plot(x_grid, -1 + 0.5 * x_grid,
        color="black", linestyle="--", label="Population regression line")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Least-squares and population regression lines")
ax.legend()
plt.show()

# (g) Polynomial regression with x and x^2.
polynomial_model = sm.OLS(y, sm.add_constant(np.column_stack((x, x**2)))).fit()
print("\n(g) Polynomial model:")
print(polynomial_model.summary())
print(f"p-value for x^2 = {polynomial_model.pvalues[2]:.4f}")

# (h)-(i): repeat the simulation with less and more noise.
def fit_simulation(error_sd):
    x_sim, eps_sim, y_sim = generate_data(error_sd)
    model_sim = sm.OLS(y_sim, sm.add_constant(x_sim)).fit()
    return x_sim, y_sim, model_sim

less_x, less_y, less_model = fit_simulation(error_sd=0.1)
more_x, more_y, more_model = fit_simulation(error_sd=1.0)

print("\n(h)-(i) Comparison of noise levels:")
comparison = pd.DataFrame({
    "Data set": ["Less noise", "Original", "More noise"],
    "Error variance": [0.01, 0.25, 1.00],
    "Estimated intercept": [less_model.params[0], original_model.params[0], more_model.params[0]],
    "Estimated slope": [less_model.params[1], original_model.params[1], more_model.params[1]],
    "R-squared": [less_model.rsquared, original_model.rsquared, more_model.rsquared],
    "Residual SD": [np.sqrt(less_model.mse_resid), np.sqrt(original_model.mse_resid), np.sqrt(more_model.mse_resid)]
})
print(comparison.round(4).to_string(index=False))

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for ax, x_sim, y_sim, model_sim, title in [
    (axes[0], less_x, less_y, less_model, "Less noise"),
    (axes[1], x, y, original_model, "Original noise"),
    (axes[2], more_x, more_y, more_model, "More noise")
]:
    ax.scatter(x_sim, y_sim, alpha=0.65)
    grid = np.linspace(x_sim.min(), x_sim.max(), 200)
    ax.plot(grid, model_sim.params[0] + model_sim.params[1] * grid, color="red")
    ax.plot(grid, -1 + 0.5 * grid, color="black", linestyle="--")
    ax.set_title(title)
    ax.set_xlabel("x")
axes[0].set_ylabel("y")
plt.tight_layout()
plt.show()

# (j) Confidence intervals for beta_0 and beta_1.
ci = pd.DataFrame({
    "Data set": ["Less noise", "Original", "More noise"],
    "beta_0 lower": [less_model.conf_int()[0, 0], original_model.conf_int()[0, 0], more_model.conf_int()[0, 0]],
    "beta_0 upper": [less_model.conf_int()[0, 1], original_model.conf_int()[0, 1], more_model.conf_int()[0, 1]],
    "beta_1 lower": [less_model.conf_int()[1, 0], original_model.conf_int()[1, 0], more_model.conf_int()[1, 0]],
    "beta_1 upper": [less_model.conf_int()[1, 1], original_model.conf_int()[1, 1], more_model.conf_int()[1, 1]]
})
print("\n(j) 95% confidence intervals:")
print(ci.round(4).to_string(index=False))
```
